```python
import numpy as np
import pandas as pd
train_data_path = "/Users/hongpite/Desktop/就业/kaggle/Titanic_data/train.csv"
test_data_path = "/Users/hongpite/Desktop/就业/kaggle/Titanic_data/test.csv"
gender_test_path = "/Users/hongpite/Desktop/就业/kaggle/Titanic_data/gender_submission.csv"
```

```python
train_data = pd.read_csv(train_data_path)
test_data = pd.read_csv(test_data_path)
gender_date = pd.read_csv(gender_test_path)
```

```python
women = train_data.loc[train_data.Sex == "female"]["Survived"]
rate_women = sum(women)/len(women)
print("% of women who survived:", rate_women)

# 我对于df，要统计里面的女性数量，则生成是否为女性的mask
# df["Sex"] == "female" 这会得到True False的mask
# 然后将其放入women = df.loc[mask]，则能挑出女性的那些数据，构建新的表格
# 然后计算len(women)得到总女性数量
# 对于存活的，我们可以women["Survived"] 得到存活的那列数据
# 那列数据正好存活是1，死亡是0，也是一种mask
# 然后sum就得到了存活的数量
```

```python
men = train_data.loc[train_data.Sex == 'male']["Survived"]
rate_men = sum(men)/len(men)

print("% of men who survived:", rate_men)
```

```python
from sklearn.ensemble import RandomForestClassifier

y = train_data["Survived"]

features = ["Pclass", "Sex", "SibSp", "Parch"]
X = pd.get_dummies(train_data[features])
X_test = pd.get_dummies(test_data[features])

# df[col]得到的是原始列
# get_dummies就是把文本类特征转化为0/1数值的独热编码

model = RandomForestClassifier(n_estimators=100, max_depth=5, random_state=1)
model.fit(X, y)
predictions = model.predict(X_test)

output = pd.DataFrame({'PassengerId': test_data["PassengerId"], 'Survived': predictions})
# 创建一个新的df，使用字典创建
# 字典的第一个key是ID，value是pd.Series 第二个是Survived，value是array

output.to_csv('/Users/hongpite/Desktop/就业/kaggle/Titanic_data/submission.csv', index=False)
print("Your submission was successfully saved!")
```
